# GHI IRR Plot Exporter

This notebook exports fast yearly PNGs and can optionally write a few zoomable HTML plots.

- Leave `INTERACTIVE_YEARS = []` to stay lightweight.
- Add years like `[2024]` when you want a zoomable HTML for that window.
- If an HTML feels heavy, set `INTERACTIVE_DOWNSAMPLE_RULE` to something like `"5min"`.


```python
from pathlib import Path
import sys

import pandas as pd

ROOT_DIR = Path.cwd().resolve().parent if Path.cwd().name == "plots" else Path.cwd().resolve()
if str(ROOT_DIR / "scripts") not in sys.path:
    sys.path.insert(0, str(ROOT_DIR / "scripts"))

from stw_metric_plot_notebook_common import (
    build_start_years,
    export_interactive_windows,
    export_static_windows,
    load_metric,
)

METRIC = "GHI"
COLOR = "#1f77b4"
WINDOW_YEARS = 1
OUTPUT_DIR = ROOT_DIR / "plots" / "ghi_plots"
STATIC_DPI = 150
INTERACTIVE_YEARS = []
INTERACTIVE_DOWNSAMPLE_RULE = None
```

```python
metric_df = load_metric(ROOT_DIR, METRIC)
start_years = build_start_years(metric_df, WINDOW_YEARS)
static_results = export_static_windows(
    metric_df,
    METRIC,
    COLOR,
    OUTPUT_DIR,
    start_years,
    years=WINDOW_YEARS,
    dpi=STATIC_DPI,
)
interactive_results = export_interactive_windows(
    metric_df,
    METRIC,
    COLOR,
    OUTPUT_DIR,
    INTERACTIVE_YEARS,
    years=WINDOW_YEARS,
    downsample_rule=INTERACTIVE_DOWNSAMPLE_RULE,
)
summary = {
    "static_pngs": len(static_results),
    "interactive_htmls": len(interactive_results),
    "interactive_years": INTERACTIVE_YEARS,
}
summary
```
