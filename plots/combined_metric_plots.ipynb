# Combined Metric Plotter

Select any combination of `GHI`, `DNI`, `DHI`, and `TEMP`, plus the calendar years to include. Running this notebook makes one combined plot with all selected metrics over the selected years.

```python
from pathlib import Path
import importlib
import sys

ROOT_DIR = Path.cwd().resolve().parent if Path.cwd().name == "plots" else Path.cwd().resolve()
if str(ROOT_DIR / "scripts") not in sys.path:
    sys.path.insert(0, str(ROOT_DIR / "scripts"))

import stw_metric_plot_notebook_common as notebook_common

notebook_common = importlib.reload(notebook_common)
available_years = notebook_common.available_years
export_combined_static_plot = notebook_common.export_combined_static_plot
load_metrics = notebook_common.load_metrics
show_combined_interactive_plot = notebook_common.show_combined_interactive_plot

# Choose from: "GHI", "DNI", "DHI", "TEMP".
SELECTED_METRICS = ["GHI", "DNI"]

# Choose any available calendar years. Non-contiguous years are allowed.
SELECTED_YEARS = [2015]

# Use "interactive" for one zoomable notebook plot or "static" for one PNG export.
PLOT_MODE = "interactive"

# Set to something like "5min" or "15min" if a full-resolution plot is too heavy.
DOWNSAMPLE_RULE = None

# Used only when PLOT_MODE = "static".
STATIC_OUTPUT_PATH = ROOT_DIR / "plots" / "combined_metric_plot.png"
STATIC_DPI = 150

# TEMP is a different unit than the irradiance metrics, so this combined plot uses raw values.
Y_AXIS_LABEL = "Raw value"
```

```python
metric_df, selected_metrics = load_metrics(ROOT_DIR, SELECTED_METRICS)
print(f"Available years: {available_years(metric_df)}")
print(f"Plotting metrics: {selected_metrics}")
print(f"Plotting years: {sorted(set(SELECTED_YEARS))}")
```

```python
if PLOT_MODE == "interactive":
    result = show_combined_interactive_plot(
        metric_df,
        selected_metrics,
        SELECTED_YEARS,
        downsample_rule=DOWNSAMPLE_RULE,
        y_axis_label=Y_AXIS_LABEL,
    )
elif PLOT_MODE == "static":
    result = export_combined_static_plot(
        metric_df,
        selected_metrics,
        SELECTED_YEARS,
        STATIC_OUTPUT_PATH,
        downsample_rule=DOWNSAMPLE_RULE,
        y_axis_label=Y_AXIS_LABEL,
        dpi=STATIC_DPI,
    )
else:
    raise ValueError('PLOT_MODE must be "interactive" or "static".')

result
```
